# Convert PCAP to Flow-Based representation and save results in CSVs
We use the NFStream tool available in Python at https://github.com/nfstream/nfstream?tab=readme-ov-file. For the conversion to proceed smoothly, put the attack traffic PCAPs in the attack directory, while the normal traffic PCAP in the normal directory.

```python
# Import libraries and set display options
import nfstream
from nfstream import NFStreamer, NFPlugin
import pandas as pd

pd.set_option('display.max_columns', 500)
pd.set_option('display.max_rows', 500)
```

```python
from pathlib import Path

p_input_normal = Path("../cAPTure/dataset_creation/raw_traffic/normal/normal_simulated/")
p_input_atk = Path("../cAPTure/dataset_creation/raw_traffic/attack/")

def get_subdirs(path):
    return [
        p for p in path.iterdir()
        if p.is_dir() and p.name != ".ipynb_checkpoints"
    ]

normal_subdirs = get_subdirs(p_input_normal)
atk_subdirs = get_subdirs(p_input_atk)
```

```python
normal_subdirs, atk_subdirs
```

## Convert Normal

```python
# Convert normal traffic to a flow-based format
import os

for p in normal_subdirs:
    out_p = "./data/flows/"+ str(p)
    os.mkdir(out_p)
    print(p)
    pcap_paths=[path for path in os.listdir(p) if path.endswith("pcap")]
    print(pcap_paths)
    for path in pcap_paths:
        print(f"converting {path} file in flow format...")
    
        df = NFStreamer(source= str(p) + "/" + path, statistical_analysis=True).to_pandas()
    
        print(f"converted {path} in flow format, size {df.size}")
    
        df.to_csv(out_p + "/" + path.removesuffix(".pcap") + ".csv", index=False)
```

## Convert Attack

```python
# Convert attack campaigns traffic to a flow-based format
import os

for p in atk_subdirs:
    out_p = "./data/flows/"+ str(p)
    os.mkdir(out_p)
    print(p)
    pcap_paths=[path for path in os.listdir(p) if path.endswith("pcap")]
    print(pcap_paths)
    for path in pcap_paths:
        print(f"converting {path} file in flow format...")
    
        df = NFStreamer(source= str(p) + "/" + path, statistical_analysis=True).to_pandas()
    
        print(f"converted {path} in flow format, size {df.size}")
    
        df.to_csv(out_p + "/" + path.removesuffix(".pcap") + ".csv", index=False)
```
